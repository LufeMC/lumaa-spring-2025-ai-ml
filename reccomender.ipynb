```python
import pandas as pd
import numpy as np 
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from typing import List, Dict
```

```python
class Recommender:
    
    def __init__(self, file:str):
        
        self.file = file
        self.df = None
        self.tfidfMatrix = None
        self.vectorizer = None
        
    #Loads the data from the file
    def LoadData(self) -> None:
       self.df = pd.read_csv(self.file)

       self.df['Genre'] = self.df['Genre'].fillna('[]')
       self.df['combined_features'] = self.df.apply(
            lambda row: f"{row['Genre']}{row['Song Name']} {row['Artist']} ",
            axis=1)
        
       print(f"Loaded {len(self.df)} songs")
        
    # Builds the tfidf Vector
    def BuildVectors(self) -> None:  
        
        maxFeature = 5000
        
        self.vectorizer = TfidfVectorizer( max_features = maxFeature, stop_words = 'english')
        self.recommendVector = self.vectorizer.fit_transform(self.df['combined_features'])
    
    # This function finds the similarity betweent he data and the user input and stores the similar values in the dictionary
    def getRecommedation(self, Userinput) -> List[Dict]:
        
        inputVector = self.vectorizer.transform([Userinput])

        similarity = cosine_similarity(inputVector, self.recommendVector)
        data = {
            'Song Name' : self.df['Song Name'],
            'Artist': self.df['Artist'],
            'Genre': self.df['Genre'],
            'Streams': self.df['Streams'],
            'Highest Charting Position': self.df['Highest Charting Position'],
            'Number of Times Charted': self.df['Number of Times Charted'],
            'Artist Followers': self.df['Artist Followers'],
            'similarity': similarity[0]
        }
    
        results = pd.DataFrame(data)
        
        TopRecommends = results.sort_values('similarity', ascending = False)

        
        recommendations = []
        for _, row in TopRecommends.iterrows():
         recommendations.append({
            'Song Name': row['Song Name'],
            'Artist': row['Artist'],
            'Genre': row['Genre'],
            'Streams': row['Streams'],
            'Highest Charting Position': row['Highest Charting Position'],
            'Number of Times Charted': row['Number of Times Charted'],
            'Artist Followers': row['Artist Followers'],
            'similarity': row['similarity']
            })
        return recommendations
```

```python
recommend = Recommender('/Users/saanv/Downloads/project spotify/spotify_dataset.csv')
```

```python
recommend.LoadData()
```

```python
recommend.BuildVectors()
```

```python
while True:
    UserInput = input("What music are you looking for?? Or type q to quit: ")
    if UserInput.lower() == 'q':
        break
    recommendation = recommend.getRecommedation(Userinput = UserInput)
    print("\nRecommend Songs:")
    for row in recommendation:
        if row['similarity'] != 0:
            print(f"\nSong: {row['Song Name']}")
            print(f"Artist: {row['Artist']}")
            print(f"Genre: {row['Genre']}")
            print(f"Streams: {row['Streams']:}")
            print(f"Chart Position: {row['Highest Charting Position']}")
            print(f"Similarity Score: {row['similarity']:.3f}")
        else:
            continue
```
